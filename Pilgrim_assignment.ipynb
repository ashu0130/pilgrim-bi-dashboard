```python
import pandas as pd

orders = pd.read_csv(r"C:\Users\asus\OneDrive\Documents\orders.csv")

orders["revenue"] = orders["quantity"] * orders["MRP"]

result = (
    orders.groupby("category")
    .agg(
        Total_Revenue=("revenue", "sum"),
        Orders=("order_id", "count")
    )
    .reset_index()
)
result["Avg_Order_Value"] = result["Total_Revenue"] / result["Orders"]
result = result.rename(columns={"category": "Category"})
result
```
